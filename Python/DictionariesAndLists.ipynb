# Dictionaries and Lists

List are used to store multiple items in a single variable

List are one of 4 built-in data types in Python, used to store collections of data, the other 3 are Tuple, Set and Dictionary.

```All with different qualities and usage.```

````python
# Example 1
# This is the easiest way to create a list

thisList = ['apple', 'banana', 'cherry']
print(thisList)
print(type(thisList))
````

## List Items

List items are ordered, changeable and allow duplicate values.

List items are indexed, the first item has index ```[0]```, the second item has index ```[1]``` etc.

````python
# This is how you can visualize the index of the list variables you create:

print(thisList[0:1])

# The [] is meant to work with the list variables
# The first value is the startpoint and the second value es the endpoint
````

````python
# Lets say we wish to display just the 3rd value

print(thisList[2:3])

# As I said, the index value will begin with the 0 value, the second value 
# is at index 1, so the third value is at index 2. The slice [2:3] gives us 
# just that third value (note: the end index is exclusive, so it stops before index 3).
````

````python
# Now lets say we wanna know how many items does the list has

print(len(thisList))
````
